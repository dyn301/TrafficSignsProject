```python
import os

# Kiểm tra working directory hiện tại
print(f"📍 Working directory: {os.getcwd()}")
print(f"\n📂 Các thư mục/file trong thư mục hiện tại:")
for item in os.listdir("."):
    print(f"  - {item}")
```

```python
import cv2
import os
import albumentations as A

transform = A.Compose([
    A.Rotate(limit=15, p=0.7),
    A.RandomBrightnessContrast(p=0.7),
    A.GaussianBlur(p=0.3),
    A.RandomScale(scale_limit=0.1, p=0.5)
])

input_root = "."  # Duyệt từ thư mục hiện tại
output_root = "dataset_aug"

total_images = 0
total_augmented = 0

for group in os.listdir(input_root):

    group_path = os.path.join(input_root, group)

    if not os.path.isdir(group_path):
        continue
    
    # Bỏ qua các thư mục không phải nhóm biển báo
    if group == "dataset_aug" or group.startswith("."):
        continue

    print(f"\n📁 Đang xử lý nhóm: {group}")

    for sign in os.listdir(group_path):

        sign_path = os.path.join(group_path, sign)

        if not os.path.isdir(sign_path):
            continue

        # tạo thư mục output giống cấu trúc input
        output_sign_path = os.path.join(output_root, group, sign)
        os.makedirs(output_sign_path, exist_ok=True)

        for img_name in os.listdir(sign_path):

            # chỉ xử lý file png, jpg, jpeg
            if not img_name.lower().endswith((".png", ".jpg", ".jpeg")):
                continue

            img_path = os.path.join(sign_path, img_name)

            image = cv2.imread(img_path)

            if image is None:
                continue

            total_images += 1

            # resize về kích thước chuẩn
            image = cv2.resize(image, (224, 224))

            # lưu ảnh gốc
            cv2.imwrite(
                os.path.join(output_sign_path, img_name),
                image
            )

            # tạo 9 ảnh augmentation
            base_name = os.path.splitext(img_name)[0]

            for i in range(9):

                augmented = transform(image=image)["image"]

                new_name = f"{base_name}_aug{i}.png"

                cv2.imwrite(
                    os.path.join(output_sign_path, new_name),
                    augmented
                )

                total_augmented += 1

        print(f"  ✓ {sign}: Xử lý xong")

print(f"\n{'='*50}")
print(f"✅ Hoàn thành!")
print(f"📊 Tổng ảnh gốc: {total_images}")
print(f"📊 Tổng ảnh augmented: {total_augmented}")
print(f"📊 Tổng ảnh (gốc + augmented): {total_images + total_augmented}")
print(f"📁 Lưu tại: {output_root}/")
print(f"{'='*50}")
```

```python
import os
import re

output_root = "dataset_aug"
renamed_count = 0

print(f"\n{'='*50}")
print(f"🔄 Đang đổi tên ảnh augmented...")
print(f"{'='*50}")

for group in os.listdir(output_root):
    group_path = os.path.join(output_root, group)
    
    if not os.path.isdir(group_path):
        continue
    
    for sign in os.listdir(group_path):
        sign_path = os.path.join(group_path, sign)
        
        if not os.path.isdir(sign_path):
            continue
        
        for img_name in os.listdir(sign_path):
            # Tìm file có dạng *_aug[0-8].png (ví dụ: 1_aug0.png)
            match = re.match(r'.+_aug(\d)\.png$', img_name, re.IGNORECASE)
            
            if match:
                aug_index = int(match.group(1))
                
                # Đổi aug0->2, aug1->3, ..., aug8->10
                new_index = aug_index + 2
                new_name = f"{new_index}.png"
                
                old_path = os.path.join(sign_path, img_name)
                new_path = os.path.join(sign_path, new_name)
                
                os.rename(old_path, new_path)
                renamed_count += 1

print(f"✅ Hoàn thành đổi tên!")
print(f"📊 Tổng file đã đổi tên: {renamed_count}")
print(f"📝 Ảnh đã được đặt tên: 2.png, 3.png, ..., 10.png")
print(f"{'='*50}")
```
