```python
import os
import sys
import numpy as np

# Ensure the project root folder is on PYTHONPATH so local src/ imports work in notebook
project_root = os.path.abspath(os.path.join('..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.dataset import load_metadata
import importlib

# Sửa import từ clip_embedding sang vit_embedding
import src.vit_embedding as vit_embedding_module
importlib.reload(vit_embedding_module)

# Đổi hàm load_clip_model thành load_vision_model
from src.vit_embedding import load_vision_model, extract_image_embeddings, save_embeddings

metadata_path = '../data/metadata.csv'
embeddings_out = '../data/image_embeddings.npy'

df = load_metadata(metadata_path)

# Gọi hàm load model của ViT
model, processor, device = load_vision_model()

# Quá trình extract giữ nguyên (vì hàm extract_image_embeddings đã được cập nhật trong vit_embedding.py)
embeddings = extract_image_embeddings(df, processor, model, device=device, batch_size=32)
print('Embeddings shape:', embeddings.shape)
save_embeddings(embeddings, embeddings_out)
print('Saved embeddings to', embeddings_out)
```
